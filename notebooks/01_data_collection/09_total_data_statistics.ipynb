# 09. 전체 데이터 통계 확인

세 개의 주요 CSV 파일을 각각 DataFrame으로 불러와서 다음을 확인합니다.

- `df.info()` — 컬럼, dtype, non-null 수
- 총 행 개수 (`len(df)`)
- 컬럼 목록 및 샘플 데이터 (`df.head()`)

### 대상 파일
1. `data/raw/youtube_channels_cleaned.csv`
2. `data/raw/channels/csv/all_channels.csv`
3. `data/raw/videos/csv/all_videos.csv`

```python
import pandas as pd

CHANNELS_CLEANED_PATH = "../../data/raw/youtube_channels_cleaned.csv"
ALL_CHANNELS_PATH = "../../data/raw/channels/csv/all_channels.csv"
ALL_VIDEOS_PATH = "../../data/raw/videos/csv/all_videos.csv"
```

## 1. youtube_channels_cleaned.csv

```python
df_cleaned = pd.read_csv(CHANNELS_CLEANED_PATH)
print(f"총 행 개수: {len(df_cleaned):,}")
print(f"총 컬럼 개수: {df_cleaned.shape[1]}")
print("컬럼 목록:", list(df_cleaned.columns))
```

## 2. all_channels.csv

```python
df_channels = pd.read_csv(ALL_CHANNELS_PATH)
print(f"총 행 개수: {len(df_channels):,}")
print(f"총 컬럼 개수: {df_channels.shape[1]}")
print("컬럼 목록:", list(df_channels.columns))
```

## 3. all_videos.csv

```python
df_videos = pd.read_csv(ALL_VIDEOS_PATH)
print(f"총 행 개수: {len(df_videos):,}")
print(f"총 컬럼 개수: {df_videos.shape[1]}")
print("컬럼 목록:", list(df_videos.columns))
```

```python
df_videos.info()
```

## 결측치 확인

```python
def show_missing(df, name):
    total = len(df)
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    print(f"=== {name} (총 {total:,}행) ===")
    if missing.empty:
        print("결측치 없음")
    else:
        result = pd.DataFrame({
            "missing": missing,
            "ratio(%)": (missing / total * 100).round(2),
        }).sort_values("missing", ascending=False)
        print(result)
    print()

show_missing(df_cleaned, "youtube_channels_cleaned.csv")
show_missing(df_channels, "all_channels.csv")
show_missing(df_videos, "all_videos.csv")
```
