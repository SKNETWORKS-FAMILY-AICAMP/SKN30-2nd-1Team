## 8. 조건을 넣어서 안전하게 조회하기

파라미터 바인딩을 사용하면 SQL 문자열을 직접 이어붙이는 방식보다 안전합니다.

```python
#이 부분부터는 확인 필요
target_video_id = "example_video_id"

one_video_df = db_utils.run_query(
    """
    SELECT
        video_id,
        comment_id,
        comment_text,
        like_count,
        published_at
    FROM comments
    WHERE video_id = %(video_id)s
    """,
    params={"video_id": target_video_id},
    display=False
)

one_video_df.head()
```
