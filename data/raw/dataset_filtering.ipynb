# dataset_long.csv, dataset_wide.csv를 youtube_channels_filtered.csv기준으로 filter하는 노트북입니다.

```python
import pandas as pd
```

```python
dataset_long = pd.read_csv("data/raw/dataset_long.csv")
dataset_wide = pd.read_csv("data/raw/dataset_wide.csv")
youtube_channels_filtered = pd.read_csv("data/raw/youtube_channels_filtered.csv")
```

```python
# channel_id 기준으로 dataset_long 필터링

filtered_dataset_long = dataset_long[

    dataset_long["channel_id"].isin(

        youtube_channels_filtered["youtube_channel_id"]

    )

]
filtered_dataset_long.head()
```

```python
# channel_id 기준으로 dataset_wide 필터링

filtered_dataset_wide = dataset_wide[

    dataset_wide["channel_id"].isin(

        youtube_channels_filtered["youtube_channel_id"]

    )

]
filtered_dataset_wide.head()
```

```python
# 저장
filtered_dataset_long.to_csv("filtered_dataset_long.csv", index=False, encoding="utf-8-sig")
print("저장 완료: filtered_dataset_long.csv")

filtered_dataset_wide.to_csv("filtered_dataset_wide.csv", index=False, encoding="utf-8-sig")
print("저장 완료: filtered_dataset_wide.csv")
```
